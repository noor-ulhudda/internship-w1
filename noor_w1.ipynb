

This dataset contains both numerical and categorical columns, and allows interesting insights about Netflix's content library.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("netflix_titles.csv")
```

```python
df.shape
```

The dataset contains thousands of rows and multiple columns, making it suitable for exploratory data analysis and machine learning.

```python
df.dtypes
```

Most columns are stored as text (object), while release_year is an integer. The data types mostly match the type of information stored.

```python
df.sample(15)
```

Random rows help inspect the dataset without bias toward the beginning.

```python
df.columns.tolist()
```

This displays all column names and helps identify naming issues.

```python
df.nunique()
```

This shows how many unique values each column contains.


```python
df["rating"].unique()
```

The dataset contains multiple content ratings such as TV-MA, PG, and R.

```python
df.corr(numeric_only=True)
```

Only numerical columns are included in the correlation matrix.

```python
df["release_year"].hist(bins=20)
plt.show()
```

Most Netflix titles were released during recent years.

```python
df.groupby("type").mean(numeric_only=True)
```

This compares average numerical values between Movies and TV Shows.

```python
df.select_dtypes(include="object")
```

This lists all categorical columns in the dataset.

```python
df["title"].str.contains("Love", case=False, na=False).sum()
```

This counts how many Netflix titles contain the word "Love".

```python
df.sort_values("release_year", ascending=False).head(5)
```

These are the newest titles available in the dataset.

```python
df[df["release_year"] > 2020]
```

This filters titles released after 2020.

```python
pd.pivot_table(df,
               values="release_year",
               index="type",
               aggfunc="mean")
```

This compares the average release year for Movies and TV Shows.

```python
df["release_year"].skew()
```

Skewness indicates whether release years are distributed symmetrically.

```python
df["release_year"].kurt()
```

Kurtosis measures how peaked the distribution is.

```python
df.cov(numeric_only=True)
```

Covariance measures how numerical variables change together.

```python
df["rating"].mode()
```

This identifies the most common Netflix content rating.

```python
df.memory_usage()
```

This shows how much memory each column occupies.
